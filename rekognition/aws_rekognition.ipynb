````python
import boto3
import os

s3 = boto3.client("s3")

bucket = "ritsky-data-lakehouse-2026"
local_folder = "/home/ritik/aws-data-engineering-lab/rekognition/image"

for root, dirs, files in os.walk(local_folder):
    for file in files:
        local_path = os.path.join(root, file)

        s3_path = os.path.relpath(local_path, local_folder)

        s3.upload_file(
            local_path,
            bucket,
            f"datasets/{s3_path}"
        )

print("Upload Complete!")
````

### **AWS Rekongnation**

````python
import pprint
import boto3 
import os 

rekognition_client = boto3.client(
            'rekognition',
            aws_access_key_id = os.getenv('AWS_ACCESS_KEY'),
            aws_secret_access_key = os.getenv('AWS_SECRET_ACCESS_KEY'),
            region_name = os.getenv("AWS_REGION")
        )
````

````python
response = rekognition_client.detect_labels(
    Image={
        "S3Object": {
            "Bucket": 'ritsky-data-lakehouse-2026',
            "Name": 'datasets/mypicscreen short.png'
        }
    },
    MaxLabels=10,
    MinConfidence=80
)

pprint.pp(response)
````

### **Detecting Language**

````python
import pprint
import boto3 
import os 

comprehend_client = boto3.client(
            'comprehend',
            aws_access_key_id = os.getenv('AWS_ACCESS_KEY'),
            aws_secret_access_key = os.getenv('AWS_SECRET_ACCESS_KEY'),
            region_name = os.getenv("AWS_REGION")
        )
````

### **Translationg Text using AWS Translate**

````python
import pprint
import boto3 
import os 

translate_client = boto3.client(
            'translate',
            aws_access_key_id = os.getenv('AWS_ACCESS_KEY'),
            aws_secret_access_key = os.getenv('AWS_SECRET_ACCESS_KEY'),
            region_name = os.getenv("AWS_REGION")
        )
````

````python
texts = [                 
    "¿Dónde está la estación de tren?",           
    "Bonjour, comment allez-vous ?",             
    "Guten Morgen! Ich lerne AWS.",              
    "Ciao, come stai oggi?",                      
    "Olá, tudo bem com você?",                       
    "नमस्ते, आपका दिन कैसा रहा?",                    
    "আপনি কেমন আছেন?",                           
    "こんにちは、今日は何を勉強しますか？",           
    "안녕하세요. 오늘 무엇을 배우나요?",               
    "你好，欢迎来到数据工程。",                         
    "Привет! Как проходит твой день?",               
    "مرحباً، كيف حالك اليوم؟",                        
    "Merhaba, bugün nasılsın?",                      
    "Sawasdee, how are you today?" 
]

response = []
for t in texts:
    res = translate_client.translate_text(
        Text=t,
        SourceLanguageCode="auto",
        TargetLanguageCode="ru"
    )
    response.append({
        "SourceText": t,
        "TranslatedText": res.get("TranslatedText"),
        "SourceLanguageCode": res.get("SourceLanguageCode")
    })

pprint.pp(response)
````
