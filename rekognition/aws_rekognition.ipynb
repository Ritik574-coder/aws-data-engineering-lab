## **AWS Rekognition** 

### **Summary**
AWS Rekognition is a fully managed computer vision service that uses deep learning to analyze images and videos. It can detect objects, faces, text, and activities, enabling developers to build intelligent applications without training or managing machine learning models.

### **Uploading file in s3 bucket for rekognation**

```python
import pprint
import boto3 
import os 

s3_client = boto3.client(
            's3',
            aws_access_key_id = os.getenv('AWS_ACCESS_KEY'),
            aws_secret_access_key = os.getenv('AWS_SECRET_ACCESS_KEY'),
            region_name = os.getenv("AWS_REGION")
        )
```

```python
response = s3_client.list_buckets()
for bucket in response['Buckets'] : 
    print(bucket['Name'])
```

```python
response = s3_client.list_objects_v2(Bucket='ritsky-data-lakehouse-2026')
for bucket in response['Contents'] : 
    pprint.pp(bucket['Key'])
```

```python
import boto3
import os

s3 = boto3.client("s3")

bucket = "ritsky-data-lakehouse-2026"
local_folder = "/home/ritik/aws-data-engineering-lab/rekognition/image"

for root, dirs, files in os.walk(local_folder):
    for file in files:
        local_path = os.path.join(root, file)

        s3_path = os.path.relpath(local_path, local_folder)

        s3.upload_file(
            local_path,
            bucket,
            f"datasets/{s3_path}"
        )

print("Upload Complete!")
```

```python
response = s3_client.list_objects(
    Bucket='ritsky-data-lakehouse-2026',
    Prefix = 'datasets/'
    )

for bucket in response['Contents'] :
    print(bucket['Key'])
```

### **AWS Rekongnation**

```python
import pprint
import boto3 
import os 

rekognition_client = boto3.client(
            'rekognition',
            aws_access_key_id = os.getenv('AWS_ACCESS_KEY'),
            aws_secret_access_key = os.getenv('AWS_SECRET_ACCESS_KEY'),
            region_name = os.getenv("AWS_REGION")
        )
```

```python
response = rekognition_client.detect_labels(
    Image={
        "S3Object": {
            "Bucket": 'ritsky-data-lakehouse-2026',
            "Name": 'datasets/mypicscreen short.png'
        }
    },
    MaxLabels=10,
    MinConfidence=80
)

pprint.pp(response)
```
