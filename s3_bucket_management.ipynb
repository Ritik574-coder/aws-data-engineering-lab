#### **1. LIST THE S3 BUCKET**

```python
from tabulate import tabulate 
import pandas as pd 
import pprint
import boto3 
import os

s3_client = boto3.client(
    's3',
    region_name=os.getenv('AWS_REGION'),
    aws_access_key_id=os.getenv('AWS_ACCESS_KEY'),
    aws_secret_access_key=os.getenv('AWS_SECRET_ACCESS_KEY')
)
```

```python
## Listing all bucket that are in s3 
response = s3_client.list_buckets()

print("The all s3 bucket list : ")
for bucket in response['Buckets'] :
    pprint.pp(f"    Bucket_Name : {bucket['Name']}")
```

#### **2. CREATE A NEW BUCKET**

```python
try : 
    bucket_name = 'the-ritsky-company-2026'

    response = s3_client.create_bucket(
        Bucket=bucket_name,
        CreateBucketConfiguration={
            "LocationConstraint" : "ap-south-1"
        }
    )
    pprint.pp(response)

except Exception as error :
    raise error
```

#### **DELETING EMPTY BUCKET**

```python
try :
    bucket_name = 'the-ritsky-company-2026'
    response = s3_client.delete_bucket(
        Bucket = bucket_name
    )
    print(f"Bucket {bucket_name} successfully deleted")

except Exception as error :
    raise error 
```

#### **CHECK WHETHER A BUCKET EXISTS AND IS ACCESSIBLE**

```python
try : 
    bucket_name = 'ritsky-bucket-2025'

    response = s3_client.head_bucket(
        Bucket=bucket_name
    )
    print(f"bucket {bucket_name} are exits and accessible") 
    pprint.pp(f"    http_status   : {response['ResponseMetadata']['HTTPStatusCode']}")
    pprint.pp(f"    bucket_arn    : {response['BucketArn']}")
    pprint.pp(f"    bucket_region : {response['BucketRegion']}")

except Exception as error :
    raise error 
```

#### **GET THE AWS REGION OF A BUCKET**

```python
try :
    bucket_name = 'ritsky-data-lakehouse-2026'

    response = s3_client.get_bucket_location(
        Bucket = bucket_name
    )
    
    pprint.pp(f"BucketLocation : {response['LocationConstraint']}")

except Exception as error :
    raise error 
```

###### **HOME WORK**
Create a bucket in region ap-south-1 using s3 boto client

```python
import boto3 
import pprint 
import os 
from botocore.exceptions import ClientError 

def creat_bucket(bucket_name: str, bucket_region: str ) -> dict : 

    """
    Function   : create_bucket(): 
        This Function Is Create an Amazone s3 Bucket In Specific AWS Region.

    Parameters : bucket_name : str , bucket_region: str

        bucket_name -> 
            The bucket name must follow Amazon S3 bucket naming
            rules and be globally unique .

        bucket_region -> 
            Write The AWS Region Where Bucket Will Be Created.

    Return : dict 
        AWS API Response Return After Successfully Bucket Created .

    Raises : ClientError 
        Raise ClientError When AWS Reject The Create Bucket Request
    """

    try :
        s3_client = boto3.client(
            's3',
            region_name = bucket_region,
            aws_access_key_id = os.getenv('AWS_ACCESS_KEY'),
            aws_secret_access_key = os.getenv('AWS_SECRET_ACCESS_KEY')
        )
        print("boto3 s3 client working successfully")

    except Exception as error :
        raise error 

    try : 
        response = s3_client.create_bucket(
            Bucket = bucket_name,
            CreateBucketConfiguration = {
                'LocationConstraint' : bucket_region
            }
        )
        pprint.pp(f"Bucket created successfully ! ")
        print(f"Bucket Name     : {bucket_name}")
        print(f"Bucket Region   : {bucket_region}")
        print(f"Bucket Location : {response.get('Location')}")
        print(f"Bucket Arn      : {response.get('BucketArn')}")
        return response

    except ClientError as error : 
        print(f"AWS error : {error}")
        raise
```

```python
bucket_name = 'ritsky-comput-service-2026'
bucket_region = 'ap-south-1'
print(
    creat_bucket(
        bucket_name=bucket_name,
        bucket_region=bucket_region
        )
    )
```
