### ========================== **SNS Topic** =========================

### **Creating SNS topic** 

```python
import pprint
import boto3 
import os 

sns_client = boto3.client(
            'sns',
            aws_access_key_id = os.getenv('AWS_ACCESS_KEY'),
            aws_secret_access_key = os.getenv('AWS_SECRET_ACCESS_KEY'),
            region_name = os.getenv("AWS_REGION")
        )
```

```python
try : 
    response = sns_client.create_topic(
        Name = "data_alerts"
    )
    pprint.pp(response)

except Exception as error :
    raise error 
```

### **Deleting sns topic in aws** 

```python
try : 
    response = sns_client.delete_topic(
        TopicArn = 'arn:aws:sns:ap-south-1:219399225390:data_alerts'
    )
    pprint.pp(response)

except Exception as error : 
    raise error 
```

### **List to all SNS**

```python
try : 
    response = sns_client.list_topics()
    pprint.pp(response)
    
except Exception as error :
    raise error 
```

### ========================== **SNS Subscriptions** =========================

```python
try : 
    response = sns_client.subscribe(
            TopicArn = 'arn:aws:sns:ap-south-1:219399225390:data_alerts',
            Protocol = 'email',
            Endpoint = 'ritik74820@gmail.com'
        )
    pprint.pp(response['SubscriptionArn'])

except Exception as error :
    raise error
```

### **List the all subscription**

```python
try : 
    response = sns_client.list_subscriptions()
    pprint.pp(response['Subscriptions'])

except Exception as error :
    raise error
```

### **Listing subscription by topic**

```python
try : 
    response = sns_client.list_subscriptions_by_topic(
        TopicArn = 'arn:aws:sns:ap-south-1:219399225390:data_alerts'
    )
    pprint.pp(response['Subscriptions'])

except Exception as error :
    raise error
```

### **Creating an SMS subscriptions using AWS SNS**

```python
response = sns_client.subscribe(
    TopicArn = 'arn:aws:sns:ap-south-1:219399225390:data_pipeline_alart',
    Protocol = 'SMS',
    Endpoint = 'YOUR_PHONE_NUMBER'
)
pprint.pp(response)
```

### **Creating email subscription using AWS SNS**

```python
response = sns_client.subscribe(
    TopicArn = 'arn:aws:sns:ap-south-1:219399225390:data_pipeline_alart',
    Protocol = 'email',
    Endpoint ='ritik74820@gmail.com'
)

pprint.pp(response)
```

### **Creating multiple subscription**

```python
emails = [
    "ritsky598@gmail.com",
    "ritik74820@gmail.com",
    "abhishek0515744@gmail.com",
    "karl.x8271@gmail.com",
    "rusa51902@gmail.com"
]

for email in emails:
    response = sns_client.subscribe(
        TopicArn='arn:aws:sns:ap-south-1:219399225390:data_pipeline_alart',
        Protocol="email",
        Endpoint=email
    )

    print(f"{email} -> {response['SubscriptionArn']}")
```

### **Deleting the subscription**

```python
try : 
    response = sns_client.unsubscribe(
        SubscriptionArn = 'arn:aws:sns:ap-south-1:219399225390:data_alerts:663d4872-76ea-42f0-9957-4482a7b991cd'
    )
    pprint.pp(response)

except Exception as error :
    raise error
```
